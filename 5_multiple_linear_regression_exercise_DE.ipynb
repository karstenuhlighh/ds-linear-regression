Ein Unternehmen, das Autositze herstellt, möchte ein Modell zur Vorhersage der Verkäufe erstellen und braucht deine Hilfe!

Im Ordner data findest du die Datei `carseats.csv`. Sie enthält 400 Beobachtungen zu den folgenden 11 Variablen:
* **Sales**:         Verkaufte Einheiten (in Tausend) an jedem Standort
* **CompPrice**:     Preis des Wettbewerbers an jedem Standort
* **Income**:        Einkommensniveau der Bevölkerung vor Ort (in Tausend Dollar)
* **Advertising**:   Lokales Werbebudget des Unternehmens an jedem Standort (in Tausend Dollar)
* **Population**:    Bevölkerungsgröße in der Region (in Tausend)
* **Price**:         Preis, den das Unternehmen an jedem Standort für Autositze verlangt
* **ShelveLoc**:     Ein Faktor mit den Stufen Bad, Good und Medium, der die Qualität des Regalplatzes für die Autositze an jedem Standort angibt
* **Age**:           Durchschnittsalter der Bevölkerung vor Ort
* **Education**:     Bildungsniveau an jedem Standort
* **Urban**:         Ein Faktor mit den Stufen No und Yes, der angibt, ob das Geschäft in einer städtischen oder ländlichen Lage liegt
* **US**:            Ein Faktor mit den Stufen No und Yes, der angibt, ob das Geschäft in den USA liegt oder nicht
